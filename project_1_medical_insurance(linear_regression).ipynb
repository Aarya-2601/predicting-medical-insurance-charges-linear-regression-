```python
#encoding the categorical features
insurance.replace({'sex':{'male':0,'female':1}},inplace=True)
insurance.replace({'smoker':{'yes':1,'no':0}},inplace=True)
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# 1. Initialize the encoder
# drop='first' avoids the dummy variable trap (perfect for Linear Regression)
# sparse_output=False returns a standard numpy array instead of a sparse matrix
encoder = OneHotEncoder(drop='first', sparse_output=False)

# 2. Fit and transform the specific column
# Note: Input must be 2D, so we use double brackets [['region']]
encoded_array = encoder.fit_transform(insurance[['region']])

# 3. Convert it back to a DataFrame with proper column names
encoded_cols = encoder.get_feature_names_out(['region'])
df_encoded = pd.DataFrame(encoded_array, columns=encoded_cols)

# 4. Combine it back with your original dataset (and drop the old text column)
insurance = pd.concat([insurance.drop(columns=['region']), df_encoded], axis=1)
```

```python
print(insurance)
```

```python
#splitting target anf features
X = insurance.drop(columns='charges',axis=1)
Y = insurance['charges']
```

```python
print(X)
```

```python
print(Y)
```

```python
#splitting the dataset and training and testing
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,random_state=2,test_size=0.2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

**Model Training**

```python
regressor = LinearRegression()
```

```python
regressor.fit(X_train,Y_train)
```

**model evaluation**

for overfitting we get a huge value of r2 score for training data and a small value for testing data

ideally it should almost be equal for training and testing data

```python
training_data_prediction=regressor.predict(X_train)
```

```python
#R squared value (lies between 0 and 1 if tends to 1 more then usually the model is good) regression score function
r2train = metrics.r2_score(Y_train,training_data_prediction)
print(r2train)
```

```python
testing_data_prediction = regressor.predict(X_test)
```

```python
r2test = metrics.r2_score(Y_test,testing_data_prediction)
print(r2test)
```

building a predictive system

```python
input_data = (31,1,25.74,0,0,0,1,0);
input_data_arr = np.asarray(input_data)
input_data_arr_reshaped = input_data_arr.reshape(1,-1) #this is basically because our model has 1070 rows in testing data but now we are giving just one row to it so it has to understand that
prediction = regressor.predict(input_data_arr_reshaped)
print(prediction)
```

```python
Y_pred = regressor.predict(X_test)
```

```python
plt.scatter(Y_test, Y_pred, color='red', alpha=0.7, label='Data Points')

perfect_line = np.linspace(min(Y_test), max(Y_test), 100)
plt.plot(perfect_line, perfect_line, color='blue', linewidth=2, label='Perfect Fit Line')

plt.xlabel('Actual Charges')
plt.ylabel('Predicted Charges')
plt.title('Actual vs Predicted (Model Performance)')
plt.legend()
```
